```python
import sys
import os

# Add the project root to sys.path
sys.path.append(os.path.abspath(".."))

import matplotlib.pyplot as plt
import pandas as pd
from src.data_loader import get_data

# Load data
df = get_data("SPY", start="2018-01-01", end="2024-12-31")
df.head()
```

```python
# Plot close price
plt.figure(figsize=(12, 6))
plt.plot(df['Close'], label='Close Price')
plt.title("SPY Close Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Calculate daily returns
df['Returns'] = df['Close'].pct_change()

# Calculate 21-day rolling volatility (std of returns)
df['Volatility'] = df['Returns'].rolling(window=21).std()

# Plot 21-Day Rolling Volatility
plt.figure(figsize=(12, 4))
plt.plot(df['Volatility'], color='orange')
plt.title("21-Day Rolling Volatility")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.grid(True)
plt.show()
```

```python
# Histogram of Daily Returns
plt.figure(figsize=(8, 4))
df['Returns'].hist(bins=50, edgecolor='black')
plt.title("Daily Returns Distribution")
plt.xlabel("Return")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```
